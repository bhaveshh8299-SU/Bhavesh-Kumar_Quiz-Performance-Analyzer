```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
# Calculate average marks
df['Average'] = df[['Quiz_1', 'Quiz_2', 'Quiz_3', 'Quiz_4']].mean(axis=1)

# Calculation of total marks
df['Total'] = df[['Quiz_1', 'Quiz_2', 'Quiz_3', 'Quiz_4']].sum(axis=1)

print(df)
# Students with avg marks above 70
result = df[df['Average'] >= 70]

print(result)
# Sort students according to average marks
df = df.sort_values(by='Average', ascending=False)

print(df)
```

## 6. Your individual project challenge
**Challenge:** Calculate average quiz score; identify top scorer and most consistent student using standard deviation.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# Overall average
# Calculate Average, Total and Standard Deviation

quiz_cols = ['Quiz_1', 'Quiz_2', 'Quiz_3', 'Quiz_4']

df['Average'] = df[quiz_cols].mean(axis=1)
df['Total'] = df[quiz_cols].sum(axis=1)
df['Std'] = df[quiz_cols].std(axis=1)

# Find top scorer
topper = df.loc[df['Average'].idxmax()]

# Find most consistent student
consistent = df.loc[df['Std'].idxmin()]

print("Top Scorer:", topper['Name'])
print("Average Marks:", topper['Average'])

print("Most Consistent Student:", consistent['Name'])
print("Standard Deviation:", consistent['Std'])

# Display updated DataFrame
display(df)
```
